```python
##################### Phase 1: Data Collection & Ingestion  #####################

import pandas as pd
from pathlib import Path

# 1. Define the exact folder path (using 'r' so Python reads the backslashes correctly)
folder_path = Path(r"C:\....... put your path here .........\raw_excel_data")

# 2. Create an empty list to store the dataframes
data_frames = []

# 3. Loop through every Excel file in the folder
for file in folder_path.glob("*.xlsx"):
    try:
        # Read only the 'Tabular Results' sheet
        df = pd.read_excel(file, sheet_name="Tabular Results")
        
        # Add a column to track which file the data came from
        df['Source_File'] = file.name
        
        # Append this dataframe to our master list
        data_frames.append(df)
        print(f"Successfully loaded: {file.name}")
        
    except Exception as e:
        print(f"Error loading {file.name}: {e}")

# 4. Combine all individual dataframes into one massive dataset
if data_frames:
    combined_df = pd.concat(data_frames, ignore_index=True)
    print("\n--- Data Ingestion Complete ---")
    print(f"Total rows in combined dataset: {len(combined_df)}")
    
    # Display the first few rows to verify
    display(combined_df.head())
else:
    print("\nNo data was found. Please check if the folder path is correct and contains .xlsx files.")
```

```python
# Save the combined dataset to a CSV file in the same folder the Jupyter Notebook
# index=False ensures that the arbitrary row numbers (0, 1, 2...) are not saved as a new column
file_name = "combined_raw_hpv_data.csv"
combined_df.to_csv(file_name, index=False)

print(f"Success! The dataset has been permanently saved as: {file_name}")
```
